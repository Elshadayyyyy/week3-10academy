```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
df = pd.read_csv("../data/SM/data/insurance.csv")
df.head()
```

```python
df[['age', 'bmi', 'children', 'charges']].hist(bins=30, figsize=(12, 8))
plt.suptitle("Distributions of Numerical Variables")
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(data=df[['age', 'bmi', 'children', 'charges']], orient='h')
plt.title("Boxplots for Numerical Variables")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))

sns.countplot(x='sex', data=df, ax=axes[0])
axes[0].set_title('Sex Distribution')

sns.countplot(x='smoker', data=df, ax=axes[1])
axes[1].set_title('Smoker Distribution')

sns.countplot(x='region', data=df, ax=axes[2])
axes[2].set_title('Region Distribution')

plt.tight_layout()
plt.show()
```

```python
num_cols = ['age', 'bmi', 'children', 'charges']

plt.figure(figsize=(8,6))
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='smoker', y='charges', data=df)
plt.title("Average Charges by Smoker Status")
plt.ylabel("Average Charges")
plt.show()

plt.figure(figsize=(6,4))
sns.barplot(x='region', y='charges', data=df)
plt.title("Average Charges by Region")
plt.ylabel("Average Charges")
plt.show()
```
